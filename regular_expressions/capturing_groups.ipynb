```python
import re
```

- ^ asserts that the match starts at the beginning of the string

- (\w*) matches "Lovelace" (the first group)

- , matches the comma.

- (\w*) matches "Ada" (the second group)

- $ asserts that the match ends at the end of the string

```python
result = re.search(r"^(\w*), (\w*)$", "Lovelace, Ada")
print(result)
# result.groups() returns a tuple containing all the captured groups from the match
print(result.groups())
# result.group(0) returns the entire match from the regular expression
print(result[0])
# result.group(1) returns the first captured group from the match - Lovelace
print(result[1])
# result.group(2) returns the second captured group from the match - Ada
print(result[2])
"{} {}".format(result[2], result[1])
```

```python
# If a match is found, the two captured groups (last name and first name) are extracted and rearranged to "first name last name" format. If no match is found (the input string doesn't follow the expected pattern), the original name is returned unchanged.
def rearrange_name(name):
    result = re.search(r"^(\w*), (\w*)$", name)
    if result is None:
        return name
    return "{} {}".format(result[2], result[1])
rearrange_name("Lovelace, Ada")
```

```python
# ^([\w \.-]*), ([\w \.-]*)$:
# The entire expression matches strings that start (^) and end ($) with two groups of characters separated by a comma and a space.
# The first group of characters (([\w \.-]*)) can contain letters, numbers, underscores, spaces, dots, and hyphens.
# After the first group, there must be a comma followed by a space (, ).
# The second group of characters (([\w \.-]*)) can also contain letters, numbers, underscores, spaces, dots, and hyphens.
def rearrange_name(name):
    result = re.search(r"^([\w \.-]*), ([\w \.-]*)$", name)
    if result == None:
        return name
    return "{} {}".format(result[2], result[1])
rearrange_name("Hopper, Grace M.")
```
