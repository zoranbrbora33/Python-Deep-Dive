```python
import re

# re.search starts scanning the string "a ghost" from the beginning.
# It looks for a sequence of 5 alphabetic characters.
# The substring "ghost" has 5 alphabetic characters, so it matches the pattern
# As a result, re.search(r"[a-zA-Z]{5}", "a ghost") would return a match object for the substring "ghost"
print(re.search(r"[a-zA-Z]{5}", "a ghost"))
```

```python
# finds all sequences of 5 letters and doesn't have to be whole words
print(re.findall(r"[a-zA-Z]{5}", "a scary ghost appeared"))
```

```python
# searches only whole words with 5 letters
re.findall(r"\b[a-zA-Z]{5}\b", "A scary ghost appeared")
```

```python
# must occur at least 5 times and at most 10 times consecutively - dont have to be whole words
print(re.findall(r"\w{5,10}", "I really like strawberries"))
```

```python
# finds all sequences of word characters (letters, digits, and underscores) that are at least 5 characters long in the given string
print(re.findall(r"\w{5,}", "I really like strawberries"))
```

```python
# find word that starts with s and between 1 and 20 chars long
print(re.search(r"s\w{,20}", "I really like strawberries"))
```
