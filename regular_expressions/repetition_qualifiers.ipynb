```python
import re
# `*` matches all characters except newline character
# `?` stops at first occurence of a character
```

```python
# match Py with any number of characters followed by n
print(re.search(r"Py.*n", "Pygmalion"))
# here happens greedy matching -> matches as much possible characters as it can
print(re.search(r"Py.*n", "Python Programming"))
# stop at first occurence of n
print(re.search(r"Py.*?n", "Python Programming"))
# greedy matching of substring
print(re.search(r"Py[a-z]*n", "Python Programming"))
print(re.search(r"Py[a-z]*n", "Pyn"))
```

```python
# designed to match sequences where one or more 'o's are followed by one or more 'l's
print(re.search(r"o+l+", "goldfish"))
print(re.search(r"o+l+", "woolly"))
print(re.search(r"o+l+", "boil"))
```

```python
# p?: Matches zero or one occurrence of the character 'p'. The ? quantifier means "zero or one".
# each: Matches the exact sequence of characters "each"
print(re.search(r"p?each", "To each their own"))
print(re.search(r"p?each", "I like peaches"))
```
