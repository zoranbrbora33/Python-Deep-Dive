`grep` is used to search text or search for patterns within files. The name "grep" stands for "global regular expression print," reflecting its ability to search for patterns using regular expressions.

`grep [options] pattern [file...]`

grep thon /usr/share/dict/words -> looking for word thon in words file

words in grep are case sensitive

If we want to match regardles of the case we need to use -i parameter:

grep -i thon /usr/share/dict/words

. wildcard matches any character:

grep l.rts /usr/share/dict/words

alerts

blurts

flirts

^ wildcard - looks for starting letters of words in a file

grep ^fruit /usr/share/dict/words

fruitcake

fruitcake

fruitier

$ wildcard - looks for words in a file that end with specific letter

grep cat$ /usr/bin/share/words

muscat

bobcat

cat
