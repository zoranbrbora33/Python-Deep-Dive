```python
import re
```

```python
# allows for lowercase and upercase p
print(re.search(r"[Pp]ython", "Python"))
```

```python
# search for string way preceded by any character
print(re.search(r"[a-z]way", "The end of the highway"))
print(re.search(r"[a-z]way", "What a way to go"))

# combine any ranges and symbols
print(re.search("cloud[a-zA-Z0-9]", "cloudy"))
print(re.search("cloud[a-zA-Z0-9]", "cloud9"))
```

```python
# looking for characters that are not letters
print(re.search(r"[^a-zA-Z]", "This is a sentence with spaces."))
# [^a-zA-Z ] matches any character that is not an uppercase letter, a lowercase letter, or a space
print(re.search(r"[^a-zA-Z ]", "This is a sentence with spaces."))

# alternative options that can get matched
print(re.search(r"cat|dog", "I like cats."))
print(re.search(r"cat|dog", "I love dogs!"))
print(re.search(r"cat|dog", "I like both dogs and cats."))

print(re.search(r"cat|dog", "I like cats."))
print(re.search(r"cat|dog", "I love dogs!"))
print(re.search(r"cat|dog", "I like both dogs and cats."))

# find all occurences of the matches
print(re.findall(r"cat|dog", "I like both dogs and cats."))
```
