```python
import re
# The re.split function will search the input string for any of the characters in the pattern [.?!] and use these characters as delimiters to split the string
re.split(r"[.?!]", "One sentence. Another one? And the last one!")
```

```python
# The pattern not only matches the delimiters but also captures them. This means that when the string is split, the delimiters are included in the resulting list as separate elements
re.split(r"([.?!])", "One sentence. Another one? And the last one!")
```

```python
re.sub(r"[\w.%+-]+@[\w.-]+", "[REDACTED]", "Received an email for go_nuts95@my.example.com")
```

```python
# [\w.%+-]+: This part matches the username portion of an email address.
# \w matches any word character (equivalent to [a-zA-Z0-9_]).
# ., %, +, - are additional characters often found in email usernames.
# + indicates that the preceding set of characters ([\w.%+-]) must appear one or more times.
# @: This matches the literal "@" symbol separating the username and domain parts of the email address.
# [\w.-]+: This part matches the domain portion of an email address.
# \w matches any word character.
# . and - are additional characters often found in domain names.
# + indicates that the preceding set of characters ([\w.-]) must appear one or more times.
re.sub(r"[\w.%+-]+@[\w.-]+", "[REDACTED]", "Received an email for go_nuts95@my.example.com")
```

```python
# ^: Anchors the match to the start of the string.
# ([\w .-]*): This is the first capturing group.
# \w matches any word character (equivalent to [a-zA-Z0-9_]).
# .- matches spaces, dots, and hyphens.
# * means zero or more occurrences of the characters within the brackets.
# , : Matches a comma followed by a space.
# ([\w .-]*): This is the second capturing group, with the same character set as the first group.
# $: Anchors the match to the end of the string.

# This specifies the format for the replacement text.
# \2 refers to the second captured group.
# \1 refers to the first captured group.
# The replacement string rearranges the matched text by swapping the two groups and inserting a space between them.
re.sub(r"^([\w .-]*), ([\w .-]*)$", r"\2 \1", "Lovelace, Ada")
```
